## Step 4 — Build `LinearRegressionGD`

Implement **all 8 methods**. No scaffolding inside the methods.

| Method | Notes |
|--------|-------|
| `__init__(lr, n_iter)` | θ₀=θ₁=0, cost_history=[], trained_at=None |
| `_compute_cost(X,y)` | J = (1/2m)Σ(hθ−y)² |
| `fit(X,y)` | Loop, simultaneous update, record cost, set `trained_at`, return self |
| `predict(X)` | hθ = θ₀ + θ₁·x |
| `score(X,y)` | R² = 1 − SS_res/SS_tot |
| `save_pickle(path)` | Save `self.__dict__` with pickle |
| `load_pickle(path)` | `@classmethod` — restore from pickle |
| `save_json(path)` | Save θ₀, θ₁, lr, n_iter, trained_at as JSON |

**Update rule (simultaneous):**
$$\theta_0 := \theta_0 - \alpha \cdot \frac{1}{m}\sum(h_\theta(x^i)-y^i)$$
$$\theta_1 := \theta_1 - \alpha \cdot \frac{1}{m}\sum(h_\theta(x^i)-y^i)\cdot x^i$$


```python
class LinearRegressionGD:
    """Batch Gradient Descent Linear Regression with save / load."""

    def __init__(self, learning_rate=0.1, n_iterations=1000):
        # YOUR CODE HERE
        self.lr = learning_rate
        self.n_iter = n_iterations
        self.theta0 = 0.0
        self.theta1 = 0.0
        self.cost_history = []
        self.trained_at = None

    def _compute_cost(self, X, y):
        # YOUR CODE HERE
        m = len(y)
        predictions = self.theta0 + self.theta1 * X
        cost = (1 / (2 * m)) * np.sum((predictions - y) **2)
        return cost

    def fit(self, X, y):
        # YOUR CODE HERE
        m = len(y)
        for _ in range(self.n_iter):
            predicitions = self.theta0 + self.theta1 * X
            error = predicitions - y

            # Simultaneous Update 
            temp0 = self.theta0 - self.lr * (2 / m) * np.sum(error)
            temp1 = self.theta1 - self.lr * (2 / m) * np.sum(error * X)

            self.theta0 = temp0
            self.theta1 = temp1

            self.cost_history.append(self._compute_cost(X, y))
        self.trained_at = datetime.now().isoformat()
        return self

    def predict(self, X):
        # YOUR CODE HERE
        return self.theta0 + self.theta1 * X

    def score(self, X, y):
        # YOUR CODE HERE
        y_pred = self.predict(X)
        ss_res = np.sum((y - y_pred) **2)
        ss_tot = np.sum((y - np.mean(y)) **2)

        return 1- ss_res / ss_tot

    def save_pickle(self, filepath):
        # YOUR CODE HERE
        with open(filepath, 'wb') as f:
            pickle.dump(self.__dict__, f)

    @classmethod
    def load_pickle(cls, filepath):
        # YOUR CODE HERE
        obj = cls()
        with open(filepath, 'rb') as f:
            obj.__dict__ = pickle.load(f)
        return obj

    def save_json(self, filepath):
        # Save: theta0, theta1, lr, n_iter, trained_at
        # YOUR CODE HERE
        data = {
            "theta0"     : self.theta0,
            "theta1"     : self.theta1,
            "lr"         : self.lr,
            "n_iter"     : self.n_iter,
            "trained_at" : self.trained_at
        }

        with open(filepath, 'w') as f:
            _json.dump(data, f)

    def __repr__(self):
        s = f"trained {self.trained_at[:10]}" if self.trained_at else "untrained"
        return f"LR_GD(lr={self.lr}, θ₀={self.theta0:.4f}, θ₁={self.theta1:.4f}, {s})"
```

## Step 6 — Experiment: Learning Rate Comparison

Your 5 learning rates: **`[0.1, 0.15, 0.3, 0.5, 1.0]`**  
Fixed: `n_iterations = 500`

**Produce all three outputs:**

1. **Printed table** — `α | Final J | Test R² | Status`  
   *(Status: `converged` if J<0.005, `diverged` if J > initial J, else `converging`)*

2. **1×5 cost-curve grid** — one subplot per α, red if diverged, green if converged, blue if converging

3. **Single overlay plot** — all 5 curves, log y-axis, legend

No scaffolding — write it yourself.


```python
# ── Experiment ──────────────────────────────────────────────────────
YOUR_LRS = [0.1, 0.15, 0.3, 0.5, 1.0]
N_ITER   = 500
results  = []   # dicts: lr, model, final_J, r2_test, status

# ── Train all models ─────────────────────────────────────────────────
# YOUR CODE HERE
for lr in YOUR_LRS:
    model = LinearRegressionGD(lr, N_ITER)
    model.fit(X_train, y_train)
    final_J = model.cost_history[-1]
    init_J  = model.cost_history[0]
    r2      = model.score(X_test, y_test)
    if final_J > init_J:
        status = "diverged"
    elif final_J < 0.005:
        status = "converged"
    else:
        status = "converging"
    results.append({"lr": lr, "model": model, "final_J": final_J, "r2_test": r2, "status": status})


# ── 1. Printed table ─────────────────────────────────────────────────
# YOUR CODE HERE

print(f"{'α':>6} | {'Final J':>10} | {'Test R²':>8} | Status")
print("-" * 45)
for r in results:
    print(f"{r['lr']:>6} | {r['final_J']:>10.6f} | {r['r2_test']:>8.4f} | {r['status']}")

# ── 2. 1×5 cost-curve grid ───────────────────────────────────────────
# YOUR CODE HERE
color_map = {"converged": "green", "diverged": "red", "converging": "blue"}
fig, axes = plt.subplots(1, 5, figsize=(18, 3))
for i, r in enumerate(results):
    c = color_map[r["status"]]
    axes[i].plot(r["model"].cost_history, color=c)
    axes[i].set_title(f"α={r['lr']} ({r['status']})")
    axes[i].set_xlabel("Iteration")
    axes[i].set_ylabel("Cost")
plt.tight_layout()
plt.show()


# ── 3. Overlay plot (log y-axis) ─────────────────────────────────────
# YOUR CODE HERE
plt.figure(figsize=(10, 4))
for r in results:
    plt.plot(r["model"].cost_history, label=f"α={r['lr']}")
plt.yscale('log')
plt.xlabel("Iteration")
plt.ylabel("Cost (log scale)")
plt.title("All Learning Rates — Cost Curves")
plt.legend()
plt.show()
```

## Step 10 — Final Questions *(write answers below each)*

**Q4.** Your GD model and sklearn produced almost the same regression line. Why do two totally different algorithms reach the same answer?

> *Answer:*,
Both algorithms minimize the same cost function (MSE), which has only one global minimum. Gradient Descent reaches that minimum slowly through many iterations, while sklearn uses the Normal Equation(XᵀX)⁻¹Xᵀy to get there in one direct step. Since the destination is the same, both arrive at nearly identical θ₀ and θ₁ values.

**Q5.** You saved the model as a `.pkl` file. Why is saving important in real ML projects?

> *Answer:*
In real projects, training is expensive and time-consuming. Without saving, we would have to retrain the model every time we need a prediction. By saving the pickle file, the trained theta values are preserved so we can load and deploy the model instantly without retraining.

**Q6.** If someone copied your notebook and ran it, would they get the same θ₀, θ₁, and predicted price as you? Why or why not?

> *Answer:*
Yes, they would get exactly the same results. This is because np.random.seed(4396) and random_state=4396 in train_test_split are both fixed. These ensure the dataset and train/test split are identical on every machine. Since Gradient Descent is deterministic, the same data always produces the same theta values and predictions.
